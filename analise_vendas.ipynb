## Projeto de Análise de Dados com Python

Neste projeto vamos analisar dados fictícios de vendas para identificar
faturamento, produtos mais vendidos, desempenho por região e evolução das vendas.


## 📌 Sobre o projeto

Este projeto tem como objetivo analisar dados fictícios de vendas
utilizando Python e Pandas.

A análise busca identificar:

- Faturamento total
- Produtos mais vendidos
- Desempenho por região
- Quantidade de produtos vendidos
- Principais insights dos dados

## 🛠️ Tecnologias contidas dentro desse projeto

- Python
- Pandas
- Matplotlib
- Google Colab


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dados = {
    "Produto": [
        "Notebook",
        "Mouse",
        "Teclado",
        "Monitor",
        "Notebook",
        "Mouse",
        "Monitor",
        "Teclado",
        "Notebook",
        "Mouse"
    ],

    "Regiao": [
        "Sudeste",
        "Sul",
        "Sudeste",
        "Nordeste",
        "Sul",
        "Sudeste",
        "Nordeste",
        "Sul",
        "Sudeste",
        "Nordeste"
    ],

    "Quantidade": [
        5,
        20,
        15,
        8,
        3,
        25,
        10,
        18,
        7,
        30
    ],

    "Preco": [
        3500,
        80,
        150,
        1200,
        3500,
        80,
        1200,
        150,
        3500,
        80
    ]
}

df = pd.DataFrame(dados)

df
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df["Faturamento"] = df["Quantidade"] * df["Preco"]

df
```

```python
faturamento_total = df["Faturamento"].sum()

faturamento_total
```

```python
vendas_produto = df.groupby("Produto")["Quantidade"].sum()

vendas_produto
```

```python
vendas_produto.plot(kind="bar")

plt.title("Quantidade Vendida por Produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade Vendida")

plt.show()
```

```python
faturamento_regiao = df.groupby("Regiao")["Faturamento"].sum()

faturamento_regiao
```

```python
faturamento_regiao.plot(kind="bar")

plt.title("Faturamento por Região")
plt.xlabel("Região")
plt.ylabel("Faturamento")

plt.show()
```

```python
plt.savefig("grafico_vendas.png")
```

```python
vendas_produto.plot(kind="bar")

plt.title("Quantidade Vendida por Produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade Vendida")

plt.savefig("grafico_vendas.png")
```
